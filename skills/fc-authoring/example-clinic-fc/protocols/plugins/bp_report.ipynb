# Blood Pressure Report
Rendered from a Jupyter notebook plugin (papermill-executed by `plugin_bp_ipynb.py`).

```python
# papermill injects this parameter (see plugin_bp_ipynb.py): a path to the server's packet JSON.
fc_packet = './fc_packet.json'
```

```python
import plotly.express as px
from tagbiopy.protocol import TagbioData, FCPacket
# Load the analysis frame from the server's packet.
fc_packet_obj = FCPacket(fc_packet)
df = TagbioData(fc_packet_obj).df
tall = df.melt(id_vars=['Department'], value_vars=['Systolic', 'Diastolic'],
               var_name='measure', value_name='mmHg')
px.box(tall, x='Department', y='mmHg', color='measure')
```
