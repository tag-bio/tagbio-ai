# Blood Pressure Report
Rendered from a Jupyter notebook plugin (papermill-executed by `plugin_bp_ipynb.py`).

```python
# papermill injects this parameter (see plugin_bp_ipynb.py): a path to the server's packet JSON.
fc_packet = './fc_packet.json'
```

```python
import plotly.express as px
from tagbiopy.protocol import TagbioData
# Load the analysis frame. fc_packet is the packet-file PATH papermill injected (see
# plugin_bp_ipynb.py); TagbioData builds the FCPacket from it internally.
df = TagbioData(fc_packet).df
# A numeric variable arrives named "<Collection>: <Variable>" (the Python SDK uses ": ");
# drop the collection prefix.
df = df.rename(columns=lambda c: c.replace('Blood Pressure: ', ''))
tall = df.melt(id_vars=['Department'], value_vars=['Systolic', 'Diastolic'],
               var_name='measure', value_name='mmHg')
px.box(tall, x='Department', y='mmHg', color='measure')
```
